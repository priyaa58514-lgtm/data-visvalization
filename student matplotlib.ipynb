```python
import matplotlib.pyplot as plt
x=[101,102,103,104,105,106,112]
y=[90,56,87,80,35,89,55]
x1=[103,104,105,108,109]
y1=[40,70,59,70,89]
plt.xlabel("roll number")
plt.ylabel("percentage")
plt.title("students performence")
plt.xlim(101,115)
plt.ylim(40,90)
plt.plot(x,y,label="line-1",color='black',linestyle='dotted',linewidth=2,marker='o',markerfacecolor='red',markersize=10)
plt.plot(x1,y1,label="line-2",alpha=0.5,color='black',linestyle='dashed',linewidth=2,marker='o',markerfacecolor='blue',markersize=10)
plt.legend()
plt.savefig("student performance.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
students=["neetu","nisha","priya","snegan","taran"]
marks=[76,34,95,90,35]
plt.bar(students,marks,color="brown",width=0.5,edgecolor="black",label="stud")
plt.legend()
plt.title("students marks in internal exam")
plt.xlabel("studnets")
plt.ylabel("marks")
plt.savefig("students marks in internal exam.png")
plt.show()
```
