```python
import pandas as pd
```

```python
df=pd.read_csv("C:/Users/Ibscai41.DGVC-MCA/Book1.csv")
df.head()
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
mn=df["sepal_width"].mean()
mn
```

```python
df["sepal_width"]=df["sepal_width"].fillna(mn)
```

```python
df.isnull().sum()
```

```python
md=df["petal_width"].median()
md
```

```python
df["petal_width"]=df["petal_width"].fillna(md)
```

```python
df.isnull().sum()
```

```python
mo=df["petal_length"].mode()[0]
mo
```

```python
df["petal_length"]=df["petal_length"].fillna(mo)
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt
x=df['petal_length']
y=df['petal_width']
plt.xlabel("petal_length")
plt.ylabel("petal_width")
plt.title("Iris data")
plt.xlim(1,10)
plt.ylim(1,2)
plt.plot(x,y,color='green',linestyle='dotted',linewidth=2,marker='o',markerfacecolor='red',markersize=5)
plt.plot(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=df['petal_length']
y=df['petal_width']
plt.xlabel("petal_length")
plt.ylabel("petal_width")
plt.title("Iris data")
plt.bar(x,y,color="brown",width=0.5,edgecolor="black",label="stud")
plt.legend()
plt.show()
```
